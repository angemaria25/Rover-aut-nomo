# Análisis de resultados — Proyecto 1 (Rover autónomo)

Este notebook lee los resultados generados por `python -m rover_ai.cli run-all`
en la carpeta `outputs/` y presenta las tablas de media y dispersión que se
reportan en el informe. No requiere dependencias externas (solo la librería
estándar).

```python
import json
from pathlib import Path

summary = json.loads(Path('../outputs/summary.json').read_text(encoding='utf-8'))
summary.keys()
```

## 1. Módulo de búsqueda (media ± desviación)

```python
for algo, metrics in summary['search'].items():
    print(f"{algo:8s} costo={metrics['cost']['mean']:.2f}±{metrics['cost']['std']:.2f}  "
          f"nodos={metrics['expanded']['mean']:.2f}±{metrics['expanded']['std']:.2f}  "
          f"tiempo_ms={metrics['runtime_ms']['mean']:.4f}±{metrics['runtime_ms']['std']:.4f}")
```

## 2. Metaheurísticas (mismo presupuesto de evaluaciones)

```python
for algo, metrics in summary['metaheuristics'].items():
    print(f"{algo:20s} objetivo={metrics['objective']['mean']:.2f}±{metrics['objective']['std']:.2f}  "
          f"costo_plan={metrics['plan_cost']['mean']:.2f}±{metrics['plan_cost']['std']:.2f}  "
          f"tiempo_ms={metrics['runtime_ms']['mean']:.2f}±{metrics['runtime_ms']['std']:.2f}")
```

## 3. Planificación: óptimo vs orden de metaheurística vs aleatorio

```python
for method, metrics in summary['planning'].items():
    print(f"{method:20s} costo_plan={metrics['plan_cost']['mean']:.2f}±{metrics['plan_cost']['std']:.2f}")
```
